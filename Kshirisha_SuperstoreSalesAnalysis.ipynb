# 4. Data Cleaning and Preprocessing

Data cleaning is an important step in the analytics process. It helps
identify missing values, duplicate records, incorrect data types, and
other inconsistencies before performing further analysis.

## 4.1 Checking Dataset Information

```python
df.info()
```

## 4.2 Checking Missing Values

Missing values can affect the accuracy of analysis. Therefore, the number
of missing values in each column is examined.

```python
missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values)
```

Duplicate records can lead to inaccurate calculations. Duplicate rows are
removed before continuing with the analysis.

## 4.5 Converting Date Columns

The Order Date and Ship Date columns are converted into datetime format to
enable time-based analysis such as monthly and yearly sales trends.

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Ship Date"] = pd.to_datetime(df["Ship Date"])

print("Date columns converted successfully.")
```

# 6. Sales Analysis

Sales analysis is performed to understand how the Superstore's revenue
changes over time and across different business dimensions such as
categories, regions, customer segments, and products.

## 6.1 Monthly Sales Trend

Analyzing sales over time helps identify changes in business performance
and seasonal patterns.

```python
# Calculate monthly sales
monthly_sales = df.groupby(
    df["Order Date"].dt.to_period("M")
)["Sales"].sum()

monthly_sales.index = monthly_sales.index.astype(str)

# Plot monthly sales
plt.figure(figsize=(12, 6))
plt.plot(monthly_sales.index, monthly_sales.values, marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

## 6.2 Yearly Sales Analysis

Year-wise sales are analyzed to understand the overall growth and changes
in sales performance across different years.

```python
yearly_sales = df.groupby(
    df["Order Date"].dt.year
)["Sales"].sum()

plt.figure(figsize=(8, 5))
plt.bar(yearly_sales.index.astype(str), yearly_sales.values)

plt.title("Yearly Sales")
plt.xlabel("Year")
plt.ylabel("Sales")
plt.tight_layout()
plt.show()
```

## 6.3 Sales by Category

Sales performance is compared across the major product categories to
identify which categories contribute the most to total sales.

```python
category_sales = df.groupby("Category")["Sales"].sum().sort_values(ascending=False)

plt.figure(figsize=(8, 5))
plt.bar(category_sales.index, category_sales.values)

plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sales")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

## 6.4 Sales by Sub-Category

Sub-category level analysis provides a more detailed view of product
performance within each major category.

```python
subcategory_sales = df.groupby("Sub-Category")["Sales"].sum().sort_values(ascending=False)

plt.figure(figsize=(10, 6))
plt.bar(subcategory_sales.index, subcategory_sales.values)

plt.title("Sales by Sub-Category")
plt.xlabel("Sub-Category")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 6.5 Regional Sales Analysis

Sales are compared across different regions to understand geographical
differences in business performance.

```python
region_sales = df.groupby("Region")["Sales"].sum().sort_values(ascending=False)

plt.figure(figsize=(8, 5))
plt.bar(region_sales.index, region_sales.values)

plt.title("Sales by Region")
plt.xlabel("Region")
plt.ylabel("Sales")
plt.tight_layout()
plt.show()
```

## 6.6 Sales by Customer Segment

Customer segment analysis helps identify which types of customers
contribute most to overall sales.

```python
segment_sales = df.groupby("Segment")["Sales"].sum().sort_values(ascending=False)

plt.figure(figsize=(8, 5))
plt.bar(segment_sales.index, segment_sales.values)

plt.title("Sales by Customer Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Sales")
plt.tight_layout()
plt.show()
```

## 6.7 Top 10 Products by Sales

The top-performing products are identified based on their total sales
contribution.

```python
top_products = (
    df.groupby("Product Name")["Sales"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(10, 6))
plt.barh(top_products.index[::-1], top_products.values[::-1])

plt.title("Top 10 Products by Sales")
plt.xlabel("Sales")
plt.ylabel("Product")
plt.tight_layout()
plt.show()
```

# 7. Profit Analysis

Profit analysis is performed to determine which categories, regions, and
products contribute positively to the overall profitability of the
business.

## 7.1 Profit by Category

```python
category_profit = (
    df.groupby("Category")["Profit"]
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(8, 5))
plt.bar(category_profit.index, category_profit.values)

plt.title("Profit by Category")
plt.xlabel("Category")
plt.ylabel("Profit")
plt.tight_layout()
plt.show()
```

## 7.2 Profit by Region

```python
region_profit = (
    df.groupby("Region")["Profit"]
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(8, 5))
plt.bar(region_profit.index, region_profit.values)

plt.title("Profit by Region")
plt.xlabel("Region")
plt.ylabel("Profit")
plt.tight_layout()
plt.show()
```

# 8. Discount and Profit Analysis

Discounts can influence both customer demand and business profitability.
This section examines the relationship between discount levels and profit.

```python
discount_profit = df.groupby("Discount")["Profit"].mean()

plt.figure(figsize=(9, 5))
plt.plot(discount_profit.index, discount_profit.values, marker="o")

plt.title("Average Profit by Discount Level")
plt.xlabel("Discount")
plt.ylabel("Average Profit")
plt.grid(True)
plt.tight_layout()
plt.show()
```

## 8.1 Relationship Between Sales and Profit

```python
plt.figure(figsize=(8, 6))

plt.scatter(
    df["Sales"],
    df["Profit"],
    alpha=0.5
)

plt.title("Sales vs Profit")
plt.xlabel("Sales")
plt.ylabel("Profit")
plt.grid(True)
plt.tight_layout()
plt.show()
```

# 9. Correlation Analysis

Correlation analysis is used to examine relationships among important
numerical variables such as Sales, Quantity, Discount, and Profit.

```python
numeric_columns = ["Sales", "Quantity", "Discount", "Profit"]

correlation = df[numeric_columns].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Between Numerical Variables")
plt.tight_layout()
plt.show()
```

# 11. Key Business Insights

The analysis results are used to identify important patterns and business
insights related to sales, profitability, products, customers, regions,
and discounts.

## 11.1 Overall Business Performance

```python
print("Overall Business Performance")
print("-" * 40)

print(f"Total Sales: {total_sales:,.2f}")
print(f"Total Profit: {total_profit:,.2f}")
print(f"Total Quantity Sold: {total_quantity:,}")
print(f"Total Orders: {total_orders:,}")
print(f"Total Customers: {total_customers:,}")

profit_margin = (total_profit / total_sales) * 100

print(f"Overall Profit Margin: {profit_margin:.2f}%")
```

## 11.2 Best-Performing Category

```python
best_sales_category = category_sales.idxmax()
highest_category_sales = category_sales.max()

best_profit_category = category_profit.idxmax()
highest_category_profit = category_profit.max()

print("Category with highest sales:")
print(best_sales_category)
print(f"Sales: {highest_category_sales:,.2f}")

print("\nCategory with highest profit:")
print(best_profit_category)
print(f"Profit: {highest_category_profit:,.2f}")
```

## 11.3 Regional Performance

```python
best_sales_region = region_sales.idxmax()
highest_region_sales = region_sales.max()

best_profit_region = region_profit.idxmax()
highest_region_profit = region_profit.max()

print("Region with highest sales:")
print(best_sales_region)
print(f"Sales: {highest_region_sales:,.2f}")

print("\nRegion with highest profit:")
print(best_profit_region)
print(f"Profit: {highest_region_profit:,.2f}")
```

## 11.4 Customer Segment Performance

```python
segment_profit = (
    df.groupby("Segment")["Profit"]
    .sum()
    .sort_values(ascending=False)
)

best_segment_sales = segment_sales.idxmax()
best_segment_profit = segment_profit.idxmax()

print("Customer segment with highest sales:")
print(best_segment_sales)
print(f"Sales: {segment_sales.max():,.2f}")

print("\nCustomer segment with highest profit:")
print(best_segment_profit)
print(f"Profit: {segment_profit.max():,.2f}")
```

## 11.6 Discount and Profit Pattern

```python
highest_discount = discount_profit.idxmax()
highest_discount_profit = discount_profit.max()

lowest_discount = discount_profit.idxmin()
lowest_discount_profit = discount_profit.min()

print("Discount level with highest average profit:")
print(f"{highest_discount:.2f}")
print(f"Average Profit: {highest_discount_profit:,.2f}")

print("\nDiscount level with lowest average profit:")
print(f"{lowest_discount:.2f}")
print(f"Average Profit: {lowest_discount_profit:,.2f}")
```
